```python
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)

    tokens = nltk.word_tokenize(text)
    stop_words = set(stopwords.words('english'))
    filtered = [w for w in tokens if w not in stop_words]

    stemmer = PorterStemmer()
    stemmed = [stemmer.stem(word) for word in filtered]

    lemmatizer = WordNetLemmatizer()
    lemmatized = [lemmatizer.lemmatize(word) for word in stemmed]

    return lemmatized
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def parse_syntax(text):
    doc = nlp(text)
    print("Tokens & POS:")
    for token in doc:
        print(token.text, token.pos_)

    print("\nDependency Parsing:")
    for token in doc:
        print(f"{token.text} --> {token.dep_} --> {token.head.text}")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def analyze_semantics(text):
    doc = nlp(text)
    print("Named Entities:")
    for ent in doc.ents:
        print(ent.text, ent.label_)

    print("\nCoreference Resolution (Simulated):")
    print("He is John. → 'He' refers to 'John'")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def extract_entities_relations(text):
    doc = nlp(text)
    print("Entities:")
    for ent in doc.ents:
        print(ent.text, ent.label_)

    print("\nRelations:")
    for token in doc:
        if token.dep_ == "ROOT":
            print(f"Relation: {token.head.text} → {token.text}")
```

```python
from textblob import TextBlob
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

def sentiment_analysis(text):
    blob = TextBlob(text)
    print("Sentiment Polarity:", blob.sentiment.polarity)

def spam_detection():
    # Sample
    texts = ['win money now', 'hello friend', 'buy now', 'how are you?']
    labels = [1, 0, 1, 0]  # 1 = spam, 0 = not spam

    vect = CountVectorizer()
    X = vect.fit_transform(texts)
    model = MultinomialNB()
    model.fit(X, labels)

    test = vect.transform(['win free prize'])
    print("Spam Prediction:", model.predict(test)[0])
```

```python
from transformers import pipeline
from googletrans import Translator
from sumy.summarizers.lsa import LsaSummarizer
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer

summarizer = pipeline("summarization")
translator = Translator()

def translate_text(text, dest='fr'):
    result = translator.translate(text, dest=dest)
    return result.text

def summarize_text(text):
    parser = PlaintextParser.from_string(text, Tokenizer("english"))
    summary = LsaSummarizer()(parser.document, 2)
    return " ".join(str(sentence) for sentence in summary)

def generate_text(prompt):
    generator = pipeline("text-generation")
    result = generator(prompt, max_length=30)
    return result[0]['generated_text']

```

```python
import speech_recognition as sr
import pyttsx3

def recognize_speech():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Speak:")
        audio = r.listen(source)
        text = r.recognize_google(audio)
        return text

def speak_text(text):
    engine = pyttsx3.init()
    engine.say(text)
    engine.runAndWait()
```

```python
from transformers import pipeline

def answer_question(question, context):
    qa_pipeline = pipeline("question-answering")
    result = qa_pipeline(question=question, context=context)
    return result['answer']
```
