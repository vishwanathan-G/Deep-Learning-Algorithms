```python
# Import Libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.preprocessing import sequence
```

```python
# Load IMDB Dataset (preprocessed)
max_features = 10000  # Vocabulary size
max_len = 500         # Maximum review length

# Load data
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(num_words=max_features)

print(x_train)

# Pad sequences to make them equal length
x_train = sequence.pad_sequences(x_train, maxlen=max_len)
x_test = sequence.pad_sequences(x_test, maxlen=max_len)

```

```python
# Build RNN Model
model = models.Sequential([
    layers.Embedding(max_features, 32),         # Word embeddings (convert integers to vectors)
    layers.SimpleRNN(32),                       # Simple Recurrent Neural Network layer
    layers.Dense(1, activation='sigmoid')       # Output layer for binary classification
])
```

```python
# Compile Model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train Model
history = model.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.2)

# Evaluate Model
test_loss, test_acc = model.evaluate(x_test, y_test)
print("\nTest Accuracy:", test_acc)
```

```python
# Predicting on test data
predictions = model.predict(x_test[5:11])

# Display Predictions
for i in range(6):
    print("\nReview", i+1)
    print("Predicted Probability of Positive Review:", predictions[i][0])
    if predictions[i][0] > 0.5:
        print("Predicted Sentiment: Positive")
    else:
        print("Predicted Sentiment: Negative")
    
    print("Actual Sentiment:", "Positive" if y_test[5 + i]==1 else "Negative")  # label of the same review as predictions[i] (x_test[5 + i])
```
